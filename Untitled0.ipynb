```python
import pandas as pd
import re

df = df.fillna("")

# =========================
# NORMALIZATION
# =========================
def normalize_name(name):
    name = str(name).lower().strip()
    name = re.sub(r"[^a-z0-9 ]", "", name)
    name = re.sub(r"\s+", " ", name)
    return name

df["normalized"] = df["agent_name"].apply(normalize_name)

# =========================
# NOISE FILTER (IMPROVED)
# =========================
noise_patterns = [
    "breadcrumb", "home page", "skip to content",
    "in this section", "url:", "menu",
    "primary about", "return to", "click here",
    "navigation", "read more"
]

def is_noise(text):
    t = str(text).lower()
    return any(p in t for p in noise_patterns) or len(t.split()) < 3

df = df[~df["agent_name"].apply(is_noise)]

# =========================
# INVALID AGENT FILTER (LESS AGGRESSIVE)
# =========================
def is_invalid_agent(name):
    name = str(name).lower()

    bad_patterns = [
        "if you", "return to", "url", "click here", "in this section"
    ]

    if any(b in name for b in bad_patterns):
        return True

    # allow short valid names like "SI-UK", "IDP"
    if len(name.strip()) < 2:
        return True

    return False

df = df[~df["agent_name"].apply(is_invalid_agent)]

# =========================
# CLEAN EMAILS / PHONES (ROBUST)
# =========================
def clean_list(x):
    if isinstance(x, list):
        items = x
    else:
        items = str(x).split(";")

    items = [i.strip() for i in items if i and str(i).strip()]
    return "; ".join(items)

df["emails"] = df["emails"].apply(clean_list)
df["phones"] = df["phones"].apply(clean_list)

# =========================
# QUALITY FILTER (FIXED LOGIC)
# =========================
def good_row(row):
    has_contact = (row["emails"].strip() != "" or row["phones"].strip() != "")
    valid_name = len(row["agent_name"].strip()) > 2
    return has_contact and valid_name

df = df[df.apply(good_row, axis=1)]

# =========================
# SAFE DEDUP (IMPROVED)
# =========================
df = df.drop_duplicates(subset=["normalized", "agent_name", "emails", "phones"])

df = df.reset_index(drop=True)

# =========================
# EXPORT CSV ONLY
# =========================
df.to_csv("clean_agents.csv", index=False)

print("DONE ✅ clean_agents.csv saved")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = df.fillna("")

# =========================
# MODERN STYLE
# =========================
plt.style.use("seaborn-v0_8-darkgrid")

PRIMARY = "#2E86AB"
SECONDARY = "#F18F01"
ACCENT = "#C73E1D"
GREEN = "#4CAF50"
PURPLE = "#7B1FA2"

# =========================
# KPI DASHBOARD (VISUAL HEADER)
# =========================
total = len(df)
unique_agents = df["agent_name"].nunique()

df["has_email"] = df["emails"].apply(lambda x: len(str(x).strip()) > 0)
df["has_phone"] = df["phones"].apply(lambda x: len(str(x).strip()) > 0)

email_count = df["has_email"].sum()
phone_count = df["has_phone"].sum()

print("\n" + "="*50)
print("📊 CRM INTELLIGENCE DASHBOARD")
print("="*50)
print(f"📦 Total Records      : {total}")
print(f"🏢 Unique Agents      : {unique_agents}")
print(f"📧 Email Coverage     : {email_count}")
print(f"📞 Phone Coverage     : {phone_count}")
print("="*50 + "\n")

# =========================
# TOP AGENTS (BAR CHART)
# =========================
plt.figure(figsize=(11,5))
top_agents = df["agent_name"].value_counts().head(10)

top_agents.plot(kind="bar", color=PRIMARY)

plt.title("🏢 Top Extracted Agents", fontsize=14, fontweight="bold")
plt.xlabel("Agent Name")
plt.ylabel("Count")
plt.xticks(rotation=40, ha="right")
plt.tight_layout()
plt.show()

# =========================
# EMAIL COVERAGE (SAFE PIE)
# =========================
email_counts = df["has_email"].value_counts().reindex([True, False], fill_value=0)

plt.figure(figsize=(6,6))

plt.pie(
    email_counts.values,
    labels=["Has Email", "No Email"],
    autopct="%1.1f%%",
    startangle=140,
    colors=[GREEN, ACCENT],
    explode=(0.05, 0)
)

plt.title("📧 Email Coverage", fontsize=13, fontweight="bold")
plt.show()

# =========================
# PHONE COVERAGE (SAFE PIE)
# =========================
phone_counts = df["has_phone"].value_counts().reindex([True, False], fill_value=0)

plt.figure(figsize=(6,6))

plt.pie(
    phone_counts.values,
    labels=["Has Phone", "No Phone"],
    autopct="%1.1f%%",
    startangle=140,
    colors=[PRIMARY, SECONDARY],
    explode=(0.05, 0)
)

plt.title("📞 Phone Coverage", fontsize=13, fontweight="bold")
plt.show()

# =========================
# COUNTRY DISTRIBUTION
# =========================
plt.figure(figsize=(11,5))

country_counts = df["country_name"].value_counts().head(10)

country_counts.plot(kind="bar", color=PURPLE)

plt.title("🌍 Country Distribution", fontsize=14, fontweight="bold")
plt.xlabel("Country")
plt.ylabel("Records")
plt.xticks(rotation=40, ha="right")
plt.tight_layout()
plt.show()

# =========================
# AGENT CLUSTERS (NORMALIZED)
# =========================
df["normalized"] = df["agent_name"].str.lower().str.strip()

plt.figure(figsize=(11,5))

norm_top = df["normalized"].value_counts().head(10)

norm_top.plot(kind="bar", color=SECONDARY)

plt.title("🧠 Agent Clusters (Normalized)", fontsize=14, fontweight="bold")
plt.xlabel("Cluster Name")
plt.ylabel("Frequency")
plt.xticks(rotation=40, ha="right")
plt.tight_layout()
plt.show()

# =========================
# DATA QUALITY VIEW (STACKED STYLE)
# =========================
labels = ["Email Present", "Phone Present", "Missing Email"]

values = [
    df["has_email"].sum(),
    df["has_phone"].sum(),
    len(df) - df["has_email"].sum()
]

plt.figure(figsize=(7,5))

plt.bar(labels, values, color=[GREEN, PRIMARY, ACCENT])

plt.title("📊 Data Quality Overview", fontsize=14, fontweight="bold")
plt.ylabel("Count")

for i, v in enumerate(values):
    plt.text(i, v + 1, str(v), ha='center')

plt.tight_layout()
plt.show()

# =========================
# FINAL SUMMARY
# =========================
print("\n✅ VISUALIZATION COMPLETE")
print("🔥 Dataset is ready for CRM / lead analysis / enrichment pipeline\n")
```
