```python
import pandas as pd
import re


df = df.fillna("")

# =========================
# NORMALIZATION
# =========================
def normalize_name(name):
    name = str(name).lower().strip()
    name = re.sub(r"[^a-z0-9 ]", "", name)
    name = re.sub(r"\s+", " ", name)
    return name

df["normalized"] = df["agent_name"].apply(normalize_name)

# =========================
# NOISE FILTER
# =========================
noise_patterns = [
    "breadcrumb", "home page", "skip to content",
    "in this section", "url:", "menu",
    "primary about", "return to", "click here",
    "navigation", "read more"
]

def is_noise(text):
    t = str(text).lower()
    return any(p in t for p in noise_patterns) or len(t.split()) < 4

df = df[~df["agent_name"].apply(is_noise)]

# =========================
# INVALID AGENT FILTER
# =========================
def is_invalid_agent(name):
    name = str(name).lower()

    bad_patterns = [
        "if you", "return to", "home page",
        "breadcrumb", "url", "click here",
        "in this section"
    ]

    if any(b in name for b in bad_patterns):
        return True

    if len(name.split()) < 2:
        return True

    return False

df = df[~df["agent_name"].apply(is_invalid_agent)]

# =========================
# CLEAN EMAILS / PHONES
# =========================
def clean_list(x):
    if isinstance(x, str):
        items = x.split(";")
    else:
        items = []

    return "; ".join([i.strip() for i in items if i and str(i).strip() != ""])

df["emails"] = df["emails"].apply(clean_list)
df["phones"] = df["phones"].apply(clean_list)

# =========================
# QUALITY FILTER
# =========================
def good_row(row):
    return (
        len(str(row["agent_name"])) > 5 and
        (row["emails"] != "" or row["phones"] != "")
    )

df = df[df.apply(good_row, axis=1)]

# =========================
# SAFE DEDUPLICATION
# =========================
df = df.drop_duplicates(subset=["normalized", "emails", "phones"])

# =========================
# RESET INDEX
# =========================
df = df.reset_index(drop=True)

# =========================
# EXPORT CLEAN DATA
# =========================
df.to_csv("clean_agents.csv", index=False)
df.to_excel("clean_agents.xlsx", index=False)

print("DONE ✅ Clean dataset saved: clean_agents.csv + clean_agents.xlsx")
```

```python
import pandas as pd
import matplotlib.pyplot as plt



df = df.fillna("")

# =========================
# GLOBAL STYLE (IMPORTANT)
# =========================
plt.style.use("seaborn-v0_8-darkgrid")

PRIMARY = "#2E86AB"
SECONDARY = "#F18F01"
ACCENT = "#C73E1D"
GREEN = "#4CAF50"

# =========================
# KPI PRINT
# =========================
total = len(df)
unique_agents = df["agent_name"].nunique()
emails = df["emails"].astype(str).apply(lambda x: len(x.strip()) > 0).sum()
phones = df["phones"].astype(str).apply(lambda x: len(x.strip()) > 0).sum()

print("📊 CRM DATA SUMMARY")
print("---------------------")
print("Total Records:", total)
print("Unique Agents:", unique_agents)
print("With Emails:", emails)
print("With Phones:", phones)

# =========================
# TOP AGENTS
# =========================
top_agents = df["agent_name"].value_counts().head(10)

plt.figure(figsize=(10,5))
top_agents.plot(kind="bar", color=PRIMARY)
plt.title("🏢 Top Agents in Dataset", fontsize=14)
plt.xlabel("Agent Name")
plt.ylabel("Count")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# =========================
# EMAIL COVERAGE
# =========================
df["has_email"] = df["emails"].astype(str).apply(lambda x: len(x.strip()) > 0)
email_counts = df["has_email"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    email_counts,
    labels=["Has Email", "No Email"],
    autopct="%1.1f%%",
    colors=[GREEN, ACCENT],
    startangle=140
)
plt.title("📧 Email Coverage")
plt.show()

# =========================
# PHONE COVERAGE
# =========================
df["has_phone"] = df["phones"].astype(str).apply(lambda x: len(x.strip()) > 0)
phone_counts = df["has_phone"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    phone_counts,
    labels=["Has Phone", "No Phone"],
    autopct="%1.1f%%",
    colors=[PRIMARY, SECONDARY],
    startangle=140
)
plt.title("📞 Phone Coverage")
plt.show()

# =========================
# COUNTRY INSIGHT
# =========================
country_counts = df["countries_supported"].astype(str).value_counts().head(10)

plt.figure(figsize=(10,5))
country_counts.plot(kind="bar", color=SECONDARY)
plt.title("🌍 Top Countries Supported")
plt.xlabel("Country Group")
plt.ylabel("Count")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# =========================
# NORMALIZED AGENTS
# =========================
norm_top = df["normalized"].value_counts().head(10)

plt.figure(figsize=(10,5))
norm_top.plot(kind="bar", color=ACCENT)
plt.title("🧠 Normalized Agent Clusters")
plt.xlabel("Agent Cluster")
plt.ylabel("Count")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# =========================
# QUALITY HEAT VIEW
# =========================
quality = [
    df["has_email"].sum(),
    df["has_phone"].sum(),
    len(df) - df["has_email"].sum()
]

labels = ["Email Present", "Phone Present", "Missing Email"]

plt.figure(figsize=(7,5))
plt.bar(labels, quality, color=[GREEN, PRIMARY, ACCENT])
plt.title("📊 Data Quality Overview")
plt.ylabel("Count")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```
